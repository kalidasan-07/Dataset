```python
!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer

# 1. Define original sentences representing our tokenized content
documents = [
    "Retrieval-Augmented Generation improves LLM responses by fetching relevant documents.",
    "Vector databases are commonly used in RAG to store and search document embeddings.",
    "A retriever fetches context, and a generator produces the final answer based on that context."
]

# 2. Initialize a pre-trained embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. Generate embeddings
embeddings = model.encode(documents)

# 4. Display details
print(f"Generated {len(embeddings)} embeddings.")
print(f"Each embedding has a dimension of: {embeddings[0].shape[0]}\n")

for i, (doc, emb) in enumerate(zip(documents, embeddings)):
    print(f"Document {i+1}: {doc}")
    print(f"Embedding Vector (First 5 values): {emb[:5]}\n")
```

```python
!pip install -q PyPDF2

import PyPDF2

# 1. Extract text from the PDF file
pdf_path = '/content/college-rules.pdf'
pdf_text = ""

with open(pdf_path, 'rb') as file:
    reader = PyPDF2.PdfReader(file)
    for page_num in range(len(reader.pages)):
        page = reader.pages[page_num]
        text = page.extract_text()
        if text:
            pdf_text += text + "\n"

print(f"Extracted {len(pdf_text)} characters from the PDF.\n")
# Display the first 500 characters to verify
print("Sample of extracted text:")
print(pdf_text[:500])
```

```python
# 2. Implement a basic chunking mechanism for RAG
def chunk_text(text, chunk_size=500, overlap=50):
    words = text.split()
    chunks = []
    for i in range(0, len(words), chunk_size - overlap):
        chunk = " ".join(words[i:i + chunk_size])
        chunks.append(chunk)
    return chunks

# Generate chunks from the extracted PDF text
document_chunks = chunk_text(pdf_text, chunk_size=150, overlap=25)

print(f"Created {len(document_chunks)} chunks from the PDF text.\n")
print("Sample of Chunk 1:")
print(document_chunks[0])
```

```python
# 3. Tokenize and embed the chunks using Sentence Transformers
from sentence_transformers import SentenceTransformer

# Using the same lightweight model as before
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')

# Generate embeddings for our PDF text chunks
chunk_embeddings = embedding_model.encode(document_chunks)

print(f"Successfully generated {len(chunk_embeddings)} embeddings for our PDF chunks.")
print(f"Embedding dimension: {chunk_embeddings[0].shape[0]}")
```
